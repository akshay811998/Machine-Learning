```python
import pickle
```

```python
loaded_model=pickle.load(open("ICP_finalized_model_randomforest.sav",'rb'))
age=int(input("Enter your age: "))
bmi = float(input("Enter your BMI: "))
children = int(input("Enter number of children: "))
sex = input("Enter your sex (M/F): ")
smoker = input("Are you a smoker? (Y/N): ")
sex_male = 1 if sex == "M" else 0
smoker_yes = 1 if smoker == "Y" else 0
result=loaded_model.predict([[age, bmi, children, sex_male, smoker_yes]])
print("Predicted Insurance Charges:", result)
```
