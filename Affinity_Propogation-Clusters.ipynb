```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
X = dataset.iloc[:, 3:5].values
```

```python
X
```

```python
from sklearn.cluster import AffinityPropagation
ap = AffinityPropagation(damping=0.5, affinity='euclidean')
ap.fit(X)
print ("No of clusters found:", len(ap.cluster_centers_indices_))
```

```python
y_ap=ap.fit_predict(X)
y_ap
```

```python
supervised = dataset
supervised
```

```python
supervised['Cluster_group'] = y_ap

supervised
```

```python
supervised.to_csv("cluster_AffinityPropagation.csv", index=False)
```

```python
dir(ap)
```

```python
centroids=ap.cluster_centers_
centroids
```

```python
from sklearn import metrics
cluster_centers_indices = ap.cluster_centers_indices_

n_clusters_ = len(cluster_centers_indices)

print("Estimated number of clusters: %d" % n_clusters_)
print("Silhouette Score: %0.3f" % metrics.silhouette_score(X, y_ap, metric='sqeuclidean'))
```

```python
supervised.columns[3]
```

```python
supervised.columns
```

```python
import seaborn as sns

colors = plt.cycler("color", plt.cm.viridis(np.linspace(0, 1, n_clusters_)))
plt.figure(figsize=(10, 6))
for k, col in zip(range(n_clusters_), colors):
    class_members = y_ap == k
    cluster_center = X[cluster_centers_indices[k]]
    
    plt.scatter(X[class_members, 0], X[class_members, 1], color=col["color"], marker='.', s=50)
    plt.scatter(cluster_center[0], cluster_center[1], color=col["color"], marker='o', s=200, edgecolors='black', linewidth=1.5)
    for x in X[class_members]:
        plt.plot([cluster_center[0], x[0]],[cluster_center[1], x[1]],color=col["color"], linewidth=0.4, alpha=0.5)
        
plt.title(f'Affinity Propagation — {n_clusters_} clusters found automatically')
plt.xlabel(supervised.columns[3])
plt.ylabel(supervised.columns[4])
plt.tight_layout()
plt.show()
```
