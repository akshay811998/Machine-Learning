```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
print ("Import Successful")
```

```python
dataset=pd.read_csv('Mall_Customers.csv')
dataset
```

```python
X=dataset.iloc[:,3:5].values
X
```

```python
from sklearn.cluster import Birch
bir=Birch(branching_factor=50, n_clusters=3, compute_labels=True)
bir.fit(X)
```

```python
y_birch=bir.fit_predict(X)
y_birch
```

```python
supervised=dataset
supervised
```

```python
supervised['Cluster_groups']=y_birch
supervised
```

```python
supervised.to_csv('Cluster-BIRCH.csv', index=False)
```
