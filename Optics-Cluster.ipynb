```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
print ("Import Success")
```

```python
dataset=pd.read_csv('Mall_Customers.csv')
dataset
```

```python
X=dataset.iloc[:,3:5].values
X
```

```python
from sklearn.cluster import OPTICS
opt=OPTICS(min_samples=5, xi=0.05, min_cluster_size=0.1)
opt.fit(X)
```

```python
y_opt=opt.fit_predict(X)
y_opt
```

```python
supervised=dataset
supervised
```

```python
supervised['Cluster_groups']=y_opt
supervised
```

```python
supervised.to_csv("Cluster-OPTICS.csv", index=False)
```
